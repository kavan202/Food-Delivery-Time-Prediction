## split and remove prefix.

```python
df["Weatherconditions"] = df["Weatherconditions"].apply(lambda x: x.split()[1] if x != np.nan else x)
```

```python
df["Time_taken(min)"] = df["Time_taken(min)"].apply(lambda x: int(x.split()[1]) if x != np.nan and x != "nan" else x)
```

## convert 'NaN'(str) into NaN {in numpy np.nan, dtype is float}

```python
df[df["Delivery_person_Age"].isnull()]
```

```python
for i in df.columns:
    df[i] = df[i].apply(lambda x: x.strip() if isinstance(x, str) else x)
    df[i] = df[i].apply(lambda x: np.nan if x == 'NaN' or x == np.nan else x)
```

## make str column into float

```python
for i in ['Delivery_person_Age','multiple_deliveries']:
    df[i] = df[i].apply(lambda x: float(x) if x != np.nan else x)
df["Delivery_person_Ratings"]= df["Delivery_person_Ratings"].apply(lambda x: float(x) if x != np.nan else x)
```

## fill nan value with mean and mod

mode could have multiple value that's why mode return list and we put here index [0]
ex: 1 1 2 2 3 5 6 4
mode is [1, 2]

```python
for i in ['Delivery_person_Age','Delivery_person_Ratings','multiple_deliveries']:
    df[i]=df[i].fillna(df[i].mean())
for i in ['Weatherconditions','Road_traffic_density','Festival']:
    df[i]=df[i].fillna(df[i].mode()[0])
```

here in rating column rating is out of range so...

```python
df["Delivery_person_Ratings"] = df["Delivery_person_Ratings"].apply(lambda x: df["Delivery_person_Ratings"].mean() if x > 5 else x)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OrdinalEncoder,OneHotEncoder
trf1 = ColumnTransformer(transformers=[
        ('oe', OrdinalEncoder(categories=["Low","Medium","High","Jam"]),["Road_traffic_density"]),
        ('onehot encoding',OneHotEncoder(drop='first', sparse_output=False),["Type_of_vehicle","Festival","City","Weatherconditions"]),
        ('scaler',StandardScaler(),df.columns)
    ],remainder='passthrough')
```

```python
pipe= Pipeline(
    [
        ("t1",trf1),
        ("t2",XGBRegressor())
    ]
)

numeric_features = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "Distance",
    "Time_Orderd",
    "Vehicle_condition",
    "multiple_deliveries",
]

trf1 = ColumnTransformer(
    transformers=[
        ("ordinal",OrdinalEncoder(categories=[["Low", "Medium", "High", "Jam"]]),
        ["Road_traffic_density"]),
        ("onehot",OneHotEncoder(
                drop="first",
                sparse_output=False,
                handle_unknown="ignore"),
            ["Type_of_vehicle", "Festival", "City", "Weatherconditions"]),
        ("scaler", StandardScaler(), numeric_features),
    ],remainder="passthrough")

pipe = Pipeline(
    [
        ("preprocessor", trf1),
        ("model", XGBRegressor(random_state=42)),
    ]
)

pipe.fit(X, y)
```

```python
p=pipe.predict(X)
r2_score(y,p)
```

```python
import joblib
joblib.dump(pipe, "pipe.pkl")
```
